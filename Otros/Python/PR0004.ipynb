### 2. Contar elementos en un diccionario

```python
productos={"Electrónica":["Smartphone","Laptop","Tablet","Auriculares","Smartwatch"],"Hogar":["Aspiradora","Microondas","Lámpara","Sofá","Cafetera"],"Ropa":["Camisa","Pantalones","Chaqueta","Zapatos","Bufanda"],"Deportes":["Pelota de fútbol","Raqueta de tenis","Bicicleta","Pesas","Cuerda de saltar"],"Juguetes":["Muñeca","Bloques de construcción","Peluche","Rompecabezas","Coche de juguete"]}
print(len(productos))
print({k:len(v) for k,v in productos.items()})
print(sum(len(v) for v in productos.values()))
```

### 3. Contador de frecuencias de palabras

### 4. Diccionario de listas

```python
asignaturas={"Matemáticas":["Ana","Carlos","Luis","María","Jorge"],"Física":["Elena","Luis","Juan","Sofía"],"Programación":["Ana","Carlos","Sofía","Jorge","Pedro"],"Historia":["María","Juan","Elena","Ana"],"Inglés":["Carlos","Sofía","Jorge","María"]}
op=int(input())
if op==1:
    a=input()
    print(asignaturas.get(a,[]))
elif op==2:
    a=input();e=input()
    asignaturas[a].append(e)
    print(asignaturas[a])
else:
    a=input();e=input()
    asignaturas[a].remove(e)
    print(asignaturas[a])
```

### 5. Diccionario invertido

### 6. Combinar dos diccionarios

```python
d1={'a':1,'b':2}
d2={'a':3,'c':4}
r=d1.copy()
for k,v in d2.items():r[k]=r.get(k,0)+v
print(r)
```

### 7. Filtrar claves y valores

```python
empleados={'Ana':1200,'Luis':900,'Juan':1500}
t=1000
print({k:v for k,v in empleados.items() if v>t})
```

### 8. Anidación de diccionarios

```python
departamentos={"Recursos Humanos":{"Ana":"Gerente","Luis":"Reclutamiento","Elena":"Asistente"},"Tecnología":{"Carlos":"Backend","María":"Frontend","Pedro":"Sistemas"}}
op=int(input())
dep=input()
if op==1:
    print(departamentos[dep])
elif op==2:
    e=input();p=input()
    departamentos[dep][e]=p
    print(departamentos[dep])
else:
    e=input()
    departamentos[dep].pop(e,None)
    print(departamentos[dep])
```

### 9. Transformación de datos

```python
estudiantes={"Ana":{"Matemáticas":8.5,"Física":9.0,"Programación":7.8}}
prom={k:sum(v.values())/len(v) for k,v in estudiantes.items()}
mat={}
for v in estudiantes.values():
    for a,n in v.items():
        mat[a]=mat.get(a,[])+[n]
matprom={k:sum(v)/len(v) for k,v in mat.items()}
print(prom)
print(matprom)
```

### 10. Diccionario de cuadrados

```python
def f(n):return {i:i*i for i in range(1,n+1)}
print(f(5))
```

### 17. Diccionario de listas invertido

```python
d={'x':[1,2],'y':[2,3]}
r={}
for k,vs in d.items():
    for v in vs:r[v]=r.get(v,[])+[k]
print(r)
```

### 22. Agrupar palabras

```python
pal=['hola','hilo','adios']
r={}
for w in pal:r.setdefault(w[0],[]).append(w)
print(r)
```

### 23. Posiciones en lista

```python
l=[2,3,2,4]
r={}
for i,v in enumerate(l):r.setdefault(v,[]).append(i)
print(r)
```

### 24. Suma acumulada

```python
l=[2,4,6]
r={}
s=0
for i,v in enumerate(l):
    s+=v
    r[i]=s
print(r)
```
