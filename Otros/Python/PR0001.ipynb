**1. Lectura de un número válido**  
Crea un programa que pida un número al usuario y siga solicitándolo hasta que el valor introducido sea realmente un número.


```python
while True:
    numero = input("Introduce un número: ")
    if numero.replace('.', '', 1).isdigit():  # permite decimales simples
        print("Número válido introducido:", numero)
        break
    else:
        print("Eso no es un número. Inténtalo de nuevo.")
```

**2. Tabla de multiplicar**  
Crea un programa que pida un número `n` y un valor `k`, y muestre los primeros `k` elementos de la tabla de multiplicar de `n`.

Ejemplo:  
Para `n = 5` y `k = 12` debe mostrar:  
5 * 1 = 5  
5 * 2 = 10  
5 * 3 = 15  
...

```python
n = int(input("Introduce n: "))
k = int(input("Introduce k: "))

for i in range(1, k + 1):
    print(n, "*", i, "=", n * i)
```

**4. Pirámide de asteriscos**  
Modifica el programa anterior para que, en lugar de un triángulo, cree una pirámide de asteriscos.  
Si el usuario introduce un número par, deberá volver a pedir el número hasta que introduzca uno impar.

```python

n = int(input("Introduce un número impar: "))
while n % 2 == 0:
    print("El número debe ser impar.")
    n = int(input("Introduce un número impar: "))

for i in range(1, n + 1, 2):     
    espacios = (n - i) // 2       
    print(" " * espacios + "*" * i)
```

**5. Número mayor y menor**

**6. Conversión de unidades**  
Crea un programa que convierta entre diferentes unidades de longitud:  
- milímetros (mm)  
- centímetros (cm)  
- metros (m)  
- kilómetros (km)

El usuario introducirá:  
1. La **cantidad**  
2. La **unidad de origen**  
3. La **unidad de destino**

El programa mostrará el valor convertido.


```python
#Tabla de conversion
factores = {
    "mm": 0.001,
    "cm": 0.01,
    "m": 1,
    "km": 1000
}

cantidad = float(input("Introduce la cantidad: "))
origen = input("Unidad de origen (mm, cm, m, km): ").lower()
destino = input("Unidad de destino (mm, cm, m, km): ").lower()

en_metros = cantidad * factores[origen]

resultado = en_metros / factores[destino]

print(f"{cantidad} {origen} son {resultado} {destino}")
```

**7. Acierta el número**  
Crea un programa que genere un número aleatorio entre 0 y 100.  
El usuario deberá intentar adivinarlo.  
Cada vez que introduzca un número, el programa indicará si el número a adivinar es **más alto** o **más bajo**.

Para generar un número aleatorio puedes usar:

`from random import *`  
`randint(a, b)` → devuelve un entero entre *a* y *b*.


```python
from random import *

numero_secreto = randint(0, 100)

while True:
    intento = int(input("Adivina el número (0-100): "))
    
    if intento < numero_secreto:
        print("Más alto.")
    elif intento > numero_secreto:
        print("Más bajo.")
    else:
        print("¡Correcto! El número era", numero_secreto)
        break
```

**8. Piedra, papel, tijeras, lagarto, spock**

Crea un programa que implemente el clásico juego:
**piedra, papel, tijeras, lagarto y spock**.

Reglas:

- Piedra aplasta a lagarto  
- Lagarto envenena a Spock  
- Spock rompe las tijeras  
- Tijeras decapitan al lagarto  
- Lagarto come al papel  
- Papel desautoriza a Spock  
- Spock vaporiza la piedra  
- Piedra rompe tijeras  
- Tijeras cortan papel  
- Papel envuelve a piedra  


```python
from random import *

opciones = ["piedra", "papel", "tijeras", "lagarto", "spock"]

gana_a = {
    "piedra": ["lagarto", "tijeras"],
    "papel": ["piedra", "spock"],
    "tijeras": ["papel", "lagarto"],
    "lagarto": ["papel", "spock"],
    "spock": ["tijeras", "piedra"]
}

puntos_usuario = 0
puntos_maquina = 0

while puntos_usuario < 5 and puntos_maquina < 5:
    usuario = input("Elige piedra, papel, tijeras, lagarto o spock: ").lower()

    if usuario not in opciones:
        print("Opción no válida. Inténtalo de nuevo.")
        continue

    maquina = choice(opciones)

    print(f"Tú: {usuario} | Máquina: {maquina}")

    if usuario == maquina:
        print("Empate.")
    elif maquina in gana_a[usuario]:
        print("¡Ganas esta ronda!")
        puntos_usuario += 1
    else:
        print("Pierdes esta ronda.")
        puntos_maquina += 1

    print(f"Marcador → Tú: {puntos_usuario} | Máquina: {puntos_maquina}")
    print("-" * 30)

if puntos_usuario == 5:
    print("🎉 ¡Has ganado la partida!")
else:
    print("😢 La máquina ha ganado la partida.")

```

**9. Secuencia de Fibonacci**

Crea un programa que genere los primeros `n` números de la secuencia de Fibonacci.

La secuencia está definida como:

- fib(0) = 1  
- fib(1) = 1  
- fib(n) = fib(n - 2) + fib(n - 1)


```python
n = int(input("¿Cuántos números de la secuencia quieres generar?: "))

a, b = 1, 1  # fib(0) y fib(1)

for i in range(n):
    print(a)
    a, b = b, a + b
```

**10. Números primos**

Crea un programa que solicite un número al usuario y le diga si es primo o no.  
Un número primo solo es divisible por 1 y por sí mismo.


```python
n = int(input("Introduce un número: "))

es_primo = True

if n <= 1:
    es_primo = False
else:
    for i in range(2, n):
        if n % i == 0:
            es_primo = False
            break

if es_primo:
    print(n, "es primo.")
else:
    print(n, "no es primo.")
```

n! = 1 * 2 * 3 * ... * n

**12. Contador de vocales y consonantes**

Crea un programa que solicite al usuario una cadena de texto  
y muestre cuántas **vocales** y cuántas **consonantes** contiene.


```python
texto = input("Introduce una cadena de texto: ").lower()

vocales_lista = "aeiou"
vocales = 0
consonantes = 0

for caracter in texto:
    if caracter.isalpha():  # solo contamos letras
        if caracter in vocales_lista:
            vocales += 1
        else:
            consonantes += 1

print("Vocales:", vocales)
print("Consonantes:", consonantes)
```

**18. Conversión de temperaturas**

Crea un programa que convierta entre **Celsius (C)**, **Fahrenheit (F)** y **Kelvin (K)**.  
El usuario introducirá primero la cantidad y la unidad de origen,  
y después la unidad de destino.


```python
temp = float(input("Introduce la temperatura: "))
origen = input("Unidad de origen (C, F, K): ").upper()
destino = input("Unidad de destino (C, F, K): ").upper()

# Convertimos primero a Celsius
if origen == "C":
    c = temp
elif origen == "F":
    c = (temp - 32) * 5/9
elif origen == "K":
    c = temp - 273.15

# Convertimos desde Celsius a la unidad destino
if destino == "C":
    resultado = c
elif destino == "F":
    resultado = c * 9/5 + 32
elif destino == "K":
    resultado = c + 273.15

print(f"{temp}°{origen} equivalen a {resultado}°{destino}")
```

**19. Generador de contraseñas**

Crea un programa que genere una contraseña aleatoria  
con la longitud que indique el usuario.  

La contraseña debe incluir:  
- letras mayúsculas  
- letras minúsculas  
- números  
- símbolos especiales


```python
from random import *
import string

longitud = int(input("Longitud de la contraseña: "))

caracteres = string.ascii_lowercase + string.ascii_uppercase + string.digits + string.punctuation

contraseña = "".join(choice(caracteres) for _ in range(longitud))

print("Contraseña generada:", contraseña)
```

**20. Ordenar lista de números**

Crea un programa que pida al usuario una lista de números separados por comas  
y muestre la lista ordenada de menor a mayor.


```python
entrada = input("Introduce números separados por comas: ")

numeros = [float(n.strip()) for n in entrada.split(",")]
numeros.sort()

print("Lista ordenada:", numeros)
```
