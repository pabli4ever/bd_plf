**1. Contar vocales y consonantes**  
Escribe una función que reciba una cadena y cuente cuántas vocales y cuántas consonantes contiene.


```python
def contar_vocales_consonantes(texto):
    texto = texto.lower()
    vocales = "aeiou"
    v = c = 0

    for t in texto:
        if t.isalpha():
            if t in vocales:
                v += 1
            else:
                c += 1
    return v, c

cadena = input("Introduce una cadena: ")
v, c = contar_vocales_consonantes(cadena)
print("Vocales:", v)
print("Consonantes:", c)
```

**5. Eliminar caracteres repetidos**  
Escribe una función que elimine los caracteres duplicados en una cadena,  
manteniendo solo la primera aparición de cada uno.


```python
def eliminar_repetidos(texto):
    resultado = ""
    for char in texto:
        if char not in resultado:
            resultado += char
    return resultado

cadena = input("Introduce una cadena: ")
print("Sin repetidos:", eliminar_repetidos(cadena))
```

**9. Frecuencia de caracteres**  
Crea una función que reciba una cadena y devuelva  
un diccionario con la frecuencia de cada carácter.


```python
def frecuencia_caracteres(texto):
    freq = {}
    for char in texto:
        freq[char] = freq.get(char, 0) + 1
    return freq

cadena = input("Introduce una cadena: ")
print(frecuencia_caracteres(cadena))
```

**11. Transformar a camelCase**

**12. Codificación RLE**  
Escribe una función que implemente la codificación RLE.  
Ejemplo: "aaron" → "a2r1o1n1".


```python
def rle_codificar(texto):
    if not texto:
        return ""
    
    resultado = ""
    actual = texto[0]
    contador = 1

    for c in texto[1:]:
        if c == actual:
            contador += 1
        else:
            resultado += actual + str(contador)
            actual = c
            contador = 1

    resultado += actual + str(contador)
    return resultado

cadena = input("Introduce una cadena: ")
print("Codificada:", rle_codificar(cadena))
```

**13. Decodificar RLE**  
Crea una función que decodifique una cadena comprimida con RLE.


```python
def rle_decodificar(cadena):
    resultado = ""
    i = 0

    while i < len(cadena):
        char = cadena[i]
        rep = int(cadena[i+1])
        resultado += char * rep
        i += 2

    return resultado

cadena = input("Introduce una cadena codificada: ")
print("Decodificada:", rle_decodificar(cadena))
```

**15. Comparar por ASCII**  
Suma los valores ASCII de cada carácter de dos cadenas  
y devuelve cuál tiene mayor valor.


```python
def valor_ascii(cadena):
    return sum(ord(c) for c in cadena)

a = input("Introduce la primera cadena: ")
b = input("Introduce la segunda cadena: ")

va = valor_ascii(a)
vb = valor_ascii(b)

if va > vb:
    print("La primera cadena tiene mayor valor ASCII.")
elif vb > va:
    print("La segunda cadena tiene mayor valor ASCII.")
else:
    print("Ambas tienen el mismo valor ASCII.")
```

**18. Rotar una cadena hacia la izquierda**
Escribe una función que rote una cadena hacia la izquierda un número dado de veces.

Ejemplo:  
"abcdef" rotado 2 veces → "cdefab".


```python
def rotar_izquierda(texto, n):
    n = n % len(texto)
    return texto[n:] + texto[:n]

cadena = input("Introduce una cadena: ")
veces = int(input("Número de rotaciones: "))
print("Resultado:", rotar_izquierda(cadena, veces))
```

**20. Reemplazar vocales**
Reemplaza todas las vocales de una cadena por un carácter dado.


```python
def reemplazar_vocales(texto, char):
    vocales = "aeiouAEIOU"
    resultado = ""

    for c in texto:
        if c in vocales:
            resultado += char
        else:
            resultado += c

    return resultado

cadena = input("Introduce una cadena: ")
char = input("Carácter de reemplazo: ")
print("Resultado:", reemplazar_vocales(cadena, char))
```

**22. Primer carácter no repetido**
Ejemplo: "aabccde" → "b".


```python
def primer_no_repetido(cadena):
    for c in cadena:
        if cadena.count(c) == 1:
            return c
    return None

cadena = input("Introduce una cadena: ")
print("Primer carácter no repetido:", primer_no_repetido(cadena))
```

**24. Prefijos y sufijos**
Ejemplo:  
"sol" →  
Prefijos: s, so, sol  
Sufijos: l, ol, sol


```python
palabra = input("Introduce una palabra: ")

print("Prefijos:")
for i in range(1, len(palabra) + 1):
    print(palabra[:i])

print("Sufijos:")
for i in range(len(palabra)):
    print(palabra[i:])
```
