**17. Dividir una lista según la media**  
Una lista con elementos ≥ media y otra con elementos < media.


```python
numeros = [10, 5, 8, 20, 15]

media = sum(numeros) / len(numeros)

mayores = [n for n in numeros if n >= media]
menores = [n for n in numeros if n < media]

print("Media:", media)
print("Mayores o iguales:", mayores)
print("Menores:", menores)
```

**18. Matriz: suma de filas y columnas**


```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Suma de filas
suma_filas = [sum(fila) for fila in matriz]

# Suma de columnas
suma_columnas = [sum(matriz[f][c] for f in range(len(matriz))) for c in range(len(matriz[0]))]

print("Suma filas:", suma_filas)
print("Suma columnas:", suma_columnas)
```

**20. Rotación de lista hacia la izquierda**
Ejemplo:  
[1, 2, 3, 4, 5] rotado 2 → [3, 4, 5, 1, 2]


```python
def rotar_lista(lista, n):
    n = n % len(lista)
    return lista[n:] + lista[:n]

numeros = [1, 2, 3, 4, 5]
print(rotar_lista(numeros, 2))
```

**21. Segunda mayor y segunda menor**


```python
def segunda_mayor_menor(lista):
    ordenada = sorted(set(lista))
    return ordenada[-2], ordenada[1]

numeros = [5, 1, 9, 3, 7]
print(segunda_mayor_menor(numeros))
```

**24. Intercalar dos listas del mismo tamaño**
Ejemplo:  
[1,3,5] y [2,4,6] → [1,2,3,4,5,6]


```python
def intercalar(a, b):
    resultado = []
    for x, y in zip(a, b):
        resultado.append(x)
        resultado.append(y)
    return resultado

print(intercalar([1,3,5], [2,4,6]))
```

**26. Frecuencia de cada elemento en una lista**


```python
def frecuencia(lista):
    dic = {}
    for n in lista:
        dic[n] = dic.get(n, 0) + 1
    return dic

numeros = [1, 2, 2, 3, 3, 3]
print(frecuencia(numeros))
```

**27. Reemplazar un número por otro en una lista**


```python
numeros = [1, 2, 3, 2, 4, 2]

a = int(input("Número a reemplazar: "))
b = int(input("Número nuevo: "))

resultado = [b if n == a else n for n in numeros]
print(resultado)
```
