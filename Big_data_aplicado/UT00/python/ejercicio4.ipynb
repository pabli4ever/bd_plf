1.- Buscar valor en un diccionario
Crea un diccionario de frutas y precios. Permite al usuario ingresar el nombre de una fruta y muestra su precio si existe en el diccionario, o un mensaje de que no está disponible en caso contrario.

```python

frutas = {
    "manzana": 1.20,
    "platano": 0.80,
    "naranja": 0.90,
    "pera": 1.10,
    "sandía": 3.50,
    "uva": 2.00
}

fruta = input("Ingrese el nombre de una fruta: ").lower()

if fruta in frutas:
    print(f"El precio de la {fruta} es ${frutas[fruta]:.2f}")
else:
    print(f"Lo sentimos, la fruta '{fruta}' no está disponible.")
```

2.- Contar elementos en un diccionario
Suponiendo un diccionario con al siguiente estructura, crea un programa que calcule cuántas categorías hay, cuántos productos tiene cada categoría y cuántos productos hay en total.

```python
productos = {
    "Electrónica": ["Smartphone", "Laptop", "Tablet", "Auriculares", "Smartwatch"],
    "Hogar": ["Aspiradora", "Microondas", "Lámpara", "Sofá", "Cafetera"],
    "Ropa": ["Camisa", "Pantalones", "Chaqueta", "Zapatos", "Bufanda"],
    "Deportes": ["Pelota de fútbol", "Raqueta de tenis", "Bicicleta", "Pesas", "Cuerda de saltar"],
    "Juguetes": ["Muñeca", "Bloques de construcción", "Peluche", "Rompecabezas", "Coche de juguete"],
}

def contar_productos(productos):
    num_categorias = len(productos)  # cantidad de categorías
    total_productos = 0
    
    print(f"Número de categorías: {num_categorias}\n")
    
    for categoria, items in productos.items():
        cantidad = len(items)
        total_productos += cantidad
        print(f"- {categoria}: {cantidad} productos")
    
    print(f"\nNúmero total de productos: {total_productos}")

contar_productos(productos)
```

3.- Contador de frecuencias de palabras
Escribe un programa que tome una frase y use un diccionario para contar la frecuencia de cada palabra.

```python

def contador_palabras(frase):
    frase = frase.lower()
    
    for signo in ",.;:!?¿¡":
        frase = frase.replace(signo," ")
    
    palabras = frase.split()
    
    frecuencias = {}
    
    for palabra in palabras:
        if palabra in frecuencias:
            frecuencias[palabra] += 1
        else:
            frecuencias[palabra] = 1
    
    return frecuencias
frase = input("Introduce una frase: ")
resultado = contador_palabras(frase)

print("Frecuencias de palabras:")
for palabra, frecuencia in resultado.items():
    print(f"{palabra}: {frecuencia}")
```

5.- Diccionario invertido
Escribe una función que tome un diccionario y devuelva otro con las claves y valores intercambiados (lo que antes eran valores ahora son claves, y viceversa).

```python
precios = {
    "orko": 9.99,
    "marine": 19.99,
    "primarca": 25.50,
    "patron": 35.00,
    "repulsor": 49.99
}

def invertir_diccionario(precios):
    invertido = {valor: clave for clave, valor in precios.items()}
    return invertido

print("Diccionario original:")
print(precios)

print("\nDiccionario invertido:")
print(invertir_diccionario(precios))
```

6.- Combinar dos diccionarios
Escribe un programa que tome dos diccionarios de productos y precios, y combine los productos comunes sumando sus precios, sin duplicar los elementos únicos.

```python
dic2 = {
    "Manzana": 1.20,
    "Pera": 1.10,
    "Naranja": 0.90,
    "Plátano": 0.75,
    "Sandía": 3.50,
    "Uva": 2.00
}

dic1 = {
    "Manzana": 0.80,   
    "Naranja": 0.10,   
    "Piña": 2.50,
    "Mango": 1.80,
    "Uva": 1.50,       
    "Melón": 3.00
}

combinado = {}

for clave in set(dic1) | set(dic2):  
    combinado[clave] = dic1.get(clave, 0) + dic2.get(clave, 0)

print("Diccionario combinado:")
print(combinado)
```

7.- Filtrar claves y valores
Dado un diccionario de empleados y salarios, filtra e imprime solo los empleados con un salario mayor a un umbral definido.

```python
# Diccionario de empleados y salarios
empleados = {
    "Ana": 1500,
    "Luis": 1800,
    "María": 2200,
    "Pedro": 1200,
    "Lucía": 2500,
    "Carlos": 2000
}

# Definir el umbral
umbral = 1800

# Filtrar empleados con salario mayor al umbral
filtrados = {}

for nombre, salario in empleados.items():
    if salario > umbral:
        filtrados[nombre] = salario

print(f"Empleados con salario mayor a {umbral}:")
print(filtrados)
```
